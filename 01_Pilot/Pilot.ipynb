# Pilot Project and Testing

## Test a CNN approach for position related features
In a spectra, the features are like pixels in a 1D image. Close by they are related and pattern can occour. Therefore a hypothesis is, that we can predict the class by comparing CNN methods after converting the spectral data to an image.

```python
# import
import numpy as np
import pandas as pd
import seaborn as sns
from tensorboard.notebook import display
from PIL import Image

from Pilot.ImageEncoder import image_encoder
```

## Data Generation Minimal Example

```python
# This could be a part of the spectra and the 1st+2nd derivative.
data = [(102, 89, 32.4, 44, 12, 0, 2, 13, 16),(12, 35, 3.4, 200, 180, 90, 18, 32, 44),(0, 0, 2, 4, 55, 19, 8, 108, 190), (10, 30, 50, 70, 90, 70, 50, 30, 10)]
```

```python
sns.lineplot(data)
```

```python
img_rgb = image_encoder(data = data, out_path='data/', shema='RGB')
img_hsv = image_encoder(data = data, out_path='data/', shema='HSV')
img_lab = image_encoder(data = data, out_path='data/', shema='LAB')
img_cmyk = image_encoder(data = data, out_path='data/', shema='CMYK')

img_L = image_encoder(data = data, out_path='data/', shema='L')
```

```python
img_rgb.save('img_rgb.png', format='PNG')
img_rgb.resize((900, 100), Image.Resampling.BOX).save('img_rgb_big.png', format='PNG')
img_hsv.convert('RGB').save('img_hsv.png', format='PNG')
img_hsv.resize((900,100), Image.Resampling.BOX).convert('RGB').save('img_hsv_big.png', format='PNG')
img_lab.convert('RGB').save('img_lab.png', format='PNG')
img_lab.resize((900,100), Image.Resampling.BOX).convert('RGB').save('img_lab_big.png', format='PNG')
img_cmyk.convert('RGB').save('img_cmyk.png', format='PNG')
img_cmyk.resize((900,100), Image.Resampling.BOX).convert('RGB').save('img_cmyk_big.png', format='PNG')

img_L.save('img_L.png', format='PNG')
img_L.resize((900,300),Image.Resampling.BOX).save('img_L_big.png', format='PNG')
```
